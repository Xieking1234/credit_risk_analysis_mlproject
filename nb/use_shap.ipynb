# Doing shap analysis

```python
import pickle
import shap
import pandas as pd
import numpy as np

```

```python
# 1. Load your trained XGBoost model and preprocessor objects
with open("../artifacts/model.pkl", "rb") as f:
    model = pickle.load(f)

with open("../artifacts/preprocessor.pkl", "rb") as f:
    preprocessor = pickle.load(f)
    scaler = preprocessor["scaler"]


# 2. Load your dataset (e.g., your training or test CSV)
df = pd.read_csv("../artifacts/train.csv")
X_raw = df.drop(columns=["loan_status"], errors="ignore")

X_scaled_array = scaler.transform(X_raw)
X_scaled = pd.DataFrame(X_scaled_array, columns=X_raw.columns)

explainer = shap.TreeExplainer(model)
shap_values = explainer(X_scaled)

print("SHAP values calculated successfully!")

# --- OPTION A: Global View (See overall feature importance across all data) ---
print("Generating global summary plot...")
shap.summary_plot(shap_values,X_scaled )
```
